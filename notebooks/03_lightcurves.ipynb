# 03 - Synthetic LSST light curves

Instantiates an sncosmo SN Ia model, draws an OpSim cadence for one sky
pointing, and builds the observed light curve (per-visit apparent magnitudes
with noise), then plots one band.

Uses `cmsne.opsim`, `cmsne.lightcurve` and `cmsne.plotting`.

```python
# If running in Colab, install the dependencies (uncomment):
# !pip install sncosmo
# !pip install git+https://github.com/LSSTDESC/OpSimSummaryV2.git

# Make the `cmsne` package importable when this notebook lives in notebooks/.
import sys, os
sys.path.insert(0, os.path.abspath(os.path.join(os.getcwd(), "..")))

import numpy as np
import sncosmo

from cmsne.opsim import initialise_opsim_summary
from cmsne.lightcurve import Transient, get_observations
from cmsne.plotting import plot_light_curve, plot_all_bands
```

## Set up the SN model

```python
# Using a built in sncosmo model
model = sncosmo.Model('salt3')

# Set the source redshift
model.set(z=0.01)

# Set the source peak (absolute) magnitude.
model.set_source_peakabsmag(-19.05, 'bessellb', 'vega')
```

## Draw a cadence and build the light curve

```python
# If using an updated opsim, there might not be detections (see output histogram).
# If so, tweak the MJD (the second input into Transient).
gen = initialise_opsim_summary(42.3, -12.5)
event = Transient(model, 61208)
lc = get_observations(event, gen, model, Show=True, use_previous=True, Verbose=True)
day_range = np.linspace(-30, 150, 111)
plot_light_curve('lsstr', model, day_range, lc)
```

```python
print(list(lc.obs_filters))
print(list(lc.obs_mag))
```

## All observed bands at once

Model curves plus observations for every band the cadence actually visited.


```python
plot_all_bands(model, day_range, lc)
```
