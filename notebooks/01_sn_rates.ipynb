# 01 - Cosmic SN Ia rates

Builds the intrinsic Type Ia supernova rate as a function of redshift from a
Madau-style star-formation history convolved with a delay-time distribution,
then applies an observability weighting to get the observed rate.

Uses `cmsne.sn_rates`.

```python
# If running in Colab, install the dependencies (uncomment):
# !pip install sncosmo
# !pip install git+https://github.com/LSSTDESC/OpSimSummaryV2.git

# Make the `cmsne` package importable when this notebook lives in notebooks/.
import sys, os
sys.path.insert(0, os.path.abspath(os.path.join(os.getcwd(), "..")))

import numpy as np
import matplotlib.pyplot as plt

from cmsne.sn_rates import Mad_son, compute_rate_grid, prob_
```

## Cosmic star-formation history (Madau)

```python
redshift = np.linspace(0, 10, 10000)
phi = Mad_son(redshift)

plt.plot(redshift, phi)
plt.xlabel('Redshift')
plt.ylabel('Number of stars formed')
plt.show()
```

## Intrinsic SN Ia rate vs redshift

```python
# compute_rate_grid convolves the star-formation history with the SN Ia
# delay-time distribution (this is the slow step).
redshift, nSN1A = compute_rate_grid(redshift)
print(nSN1A[1])

# look into local comoving rate
plt.plot(redshift, nSN1A)
plt.xlabel('Redshift')
plt.ylabel('Number of Type 1A Supernovae')
plt.grid(True)
plt.show()
print(np.argmin(np.abs(redshift - 1.5)))
```

## Observed SN Ia rate (with observability weighting)

```python
nSN1A_observed = nSN1A * prob_(redshift)

plt.plot(redshift, nSN1A_observed)
plt.xlabel('Redshift')
plt.ylabel('Number of observed Type 1A Supernovae')
plt.grid(True)
plt.show()
```
